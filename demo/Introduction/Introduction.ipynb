## Create the ASE.Atoms object
Using $H_2$ as our example

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent.parent))
from ase import Atoms

H2 = Atoms(symbols = ['H', 'H'], 
           positions=[[0, 0, 0], [0, 0, 0.74]])
```

## Create a calculator

```python
from aimnet.calculators import AIMNet2ASE, AIMNet2Calculator
# from mace.calculators import MACECalculator

# Here we use the 'aimnet2' alias
calculator = AIMNet2ASE('aimnet2', charge = 0)
# Other aliases for AIMNet2 can be found in the documentation
# https://github.com/isayevlab/aimnetcentral/
```

<div style="border-radius:2px; overflow:hidden; border:1px solid #dee2e6;">
  <div style="font-weight:600; padding:10px 14px; border-bottom:1px solid #c3e6cb; background-color:#d4edda; color:#155724;">AIMNet2 device selection</div>
  <div style="padding:14px;">To choose a device explicitly, construct the base calculator first:<pre><code>
from aimnet.calculators import AIMNet2ASE, AIMNet2Calculator
base_calc = AIMNet2Calculator('aimnet2', device='cpu')
calculator = AIMNet2ASE(base_calc, charge=0)</code></pre></div>
</div>

<div style="border-radius:2px; overflow:hidden; border:1px solid #dee2e6;">
  <div style="font-weight:600; padding:10px 14px; border-bottom:1px solid #bee5eb; background-color:#d1ecf1; color:#0c5460;">MACE alternative</div>
  <div style="padding:14px;">The same ASE attachment pattern works with MACE:<pre><code>
from mace.calculators import MACECalculator
calculator = MACECalculator(model_paths='/path/to/model', device='cpu', default_dtype='float64')
</code></pre></div>
</div>

## Attach the calculator to the ASE.Atoms object

```python
H2.calc = calculator
```

## Some available properties

```python
# Potential energy
print("Potential energy:")
print(H2.get_potential_energy())
print()

# Forces
print("Forces:")
print(H2.get_forces())
print()

# Hessian
print("Hessian:")
print(H2.calc.get_hessian())
print()
```

<div role="alert" class="alert alert-info">ASE reports energy in eV and forces in eV/&#197;. The Hessian is an energy second derivative in eV/&#197;&#178;; check units before comparing models or references.</div>

```python
# Optimize
from ase.optimize import BFGS
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd().parent.parent))
from demo_tools.optimization import optimize_ase

opt = optimize_ase(H2, optimizer=BFGS, fmax=0.00001)

print(H2.get_potential_energy())
print(H2.get_forces())
print(H2.calc.get_hessian())
```
