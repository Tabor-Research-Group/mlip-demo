# Introduction

Each exercise in this workshop follows a simple template  
    &emsp;(1) Build/load into an ASE.Atoms object  
    &emsp;(2) Create a calculator  
    &emsp;&emsp;&emsp;  aimnet2, mace  
    &emsp;(3) Attach the calculator to the ASE.Atoms object  
    &emsp;(4) Optimize the structure and compute properties

There will be some variation in how to approach different systems.  Such as, adding constraints to optimizations and allowing unit cell vectors to be optimized with atomic positions.

## Create the ASE.Atoms object
Using $H_2$ as our example

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent.parent))
from ase import Atoms

H2 = Atoms(symbols = ['H', 'H'], 
           positions=[[0, 0, 0], [0, 0, 0.74]])
from demo_tools.jhtml import JHTML
```

## Create a calculator

```python
from aimnet.calculators import AIMNet2ASE, AIMNet2Calculator
# from mace.calculators import MACECalculator

# Here we use the 'aimnet2' alias
calculator = AIMNet2ASE('aimnet2', charge = 0)
# Other aliases for AIMNet2 can be found in the documentation
# https://github.com/isayevlab/aimnetcentral/
```

```python
JHTML.JCard(
    "To choose a device explicitly, construct the base calculator first:",
    JHTML.Pre("from aimnet.calculators import AIMNet2ASE, AIMNet2Calculator\nbase_calc = AIMNet2Calculator('aimnet2', device='cpu')\ncalculator = AIMNet2ASE(base_calc, charge=0)"),
    title="AIMNet2 device selection",
    kind="success",
)
```

```python
JHTML.JCard(
    "The same ASE attachment pattern works with MACE:",
    JHTML.Pre("from mace.calculators import MACECalculator\ncalculator = MACECalculator(model_paths='/path/to/model', device='cpu', default_dtype='float64')"),
    title="MACE alternative",
    kind="info",
)
```

## Attach the calculator to the ASE.Atoms object

```python
H2.calc = calculator
```

## Some available properties

```python
# Potential energy
print("Potential energy:")
print(H2.get_potential_energy())
print()

# Forces
print("Forces:")
print(H2.get_forces())
print()

# Hessian
print("Hessian:")
print(H2.calc.get_hessian())
print()
```

```python
JHTML.JAlert(
    "ASE reports energy in eV and forces in eV/Å. The Hessian is an energy second derivative in eV/Å²; check units before comparing models or references.",
    kind="info",
)
```

```python
# Optimize
from ase.optimize import BFGS
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd().parent.parent))
from demo_tools.optimization import optimize_ase

opt = optimize_ase(H2, optimizer=BFGS, fmax=0.00001)

print(H2.get_potential_energy())
print(H2.get_forces())
print(H2.calc.get_hessian())
```

```python
JHTML.JAlert(
    f"BFGS finished after {opt.nsteps} steps. " + ("The force threshold was met." if opt.converged() else "The force threshold was not met."),
    kind="success" if opt.converged() else "warning",
)
```
