# Optional planning exercise: a Diels-Alder transition state
**Status: discussion extension, not a runnable reaction calculation.** The original notebook contained
only a placeholder. This version turns that placeholder into a guided design exercise without inventing
reactant structures or validated results.

Complete the SN2 exercise first. Use its workflow as a starting point, but do not reuse its charge,
reaction coordinate, or atom indices for another chemical system.

## 1. Define the chemical question
Choose and draw a specific diene and dienophile. State total charge, spin state, and the stereochemical
outcome to investigate. Identify both forming bonds and any competing approaches.

## 2. Design a guess
Would a single bond-distance scan describe both forming bonds adequately? Sketch a two-coordinate
surface and consider synchronous and asynchronous paths. Plan how you would obtain a plausible
reactant complex and avoid close contacts.

## 3. Choose and validate a model
Check element coverage and suitability for bond rearrangements. Specify what reference energies,
forces, or structures you would use to evaluate the model. Element coverage alone is not sufficient.

## 4. Define acceptance criteria
A converged saddle should have one meaningful imaginary mode involving the intended bond changes.
Follow both reaction-path directions and inspect connectivity and stereochemistry at the endpoints.
Separate an electronic potential-energy barrier from a free-energy barrier.

## Deliverable
Write a short calculation plan stating the system, electronic state, two forming-bond coordinates,
initial-guess strategy, model-validation data, and criteria for accepting or rejecting the result.
This extension has no supplied numerical answer and is not required for the main workshop.
